```python
df = pd.read_csv('encuestaIA.csv')

print(df.head())
```

```python
df = df.drop(columns=['Marca temporal'])
```

```python
df = pd.get_dummies(df, columns=
                    ['1. ¿Crees tú que la inteligencia artificial ha sido una herramienta que ha aportado benéficamente a la sociedad?'], prefix='beneficioIA')
```

```python
df.rename(columns={
    '2. ¿Cuánto conocimiento tienes sobre el desface que tiene IA sobre los códigos éticos?': 'conocimientoDesfaseIA',
    '3. ¿Cuánto conocimiento tienes sobre la importancia de la ética en el uso de IA?': 'conocimientoEticaIA',
    '4. ¿Crees tú que es de suma importancia los códigos éticos al momento de utilizar IA?': 'importanciaCodigosIA',
    '5. ¿Qué tan importante es para ti la ética cuando utilizas IA?': 'importanciaEticaIA',
    '6. ¿ Crees que es importante establecer parámetros  al momento de utilizar IA para que esta herramienta sea benéfica y no perjudicial para tu desarrollo académico?': 'parametrosUsoIA',
    '7. ¿Estás de acuerdo en que  la inteligencia artificial tenga regulaciones éticas estrictas para evitar el plagio en los trabajos académicos?': 'regularIAPlagio',
    '8. ¿Estás de acuerdo en que  se penalice a la persona que utilice IA sin tomar en cuenta los códigos éticos para el uso benéfico  máximo de esta herramienta?': 'penalizarIACodigos'
}, inplace=True)
```

```python
mapping_preg2 = {
    "Muy poco": 0,
    "Poco": 1,
    "Regular": 2,
    "Bueno": 3,
    "Muy bueno": 4
}
df['conocimientoDesfaseIA'] = df['conocimientoDesfaseIA'].map(mapping_preg2)
df['conocimientoEticaIA'] = df['conocimientoEticaIA'].map(mapping_preg2)
```

```python
mapping_preg4 = {
    "No es importante": 0,
    "Poco importante": 1,
    "Neutral": 2,
    "Importante": 3,
    "Muy importante": 4
}
df['importanciaCodigosIA'] = df['importanciaCodigosIA'].map(mapping_preg4)
df['importanciaEticaIA'] = df['importanciaEticaIA'].map(mapping_preg4)
df['parametrosUsoIA'] = df['parametrosUsoIA'].map(mapping_preg4)
```

```python
mapping_preg7 = {
    "Muy en desacuerdo": 0,
    "En desacuerdo": 1,
    "Neutral": 2,
    "De acuerdo": 3,
    "Totalmente de acuerdo": 4
}
df['regularIAPlagio'] = df['regularIAPlagio'].map(mapping_preg7)
df['penalizarIACodigos'] = df['penalizarIACodigos'].map(mapping_preg7)
```

```python
print(df.head())
```

```python
df.to_csv('encuestaIA_procesada.csv', index=False, encoding='utf-8')
```
